# Embed PNGs and GIFs directly into a markdown cell

This Jupyter notebook includes Python code blocks which transform PNGs and GIFs to renderable strings in markdown cells. After running a code block, paste the outcome string into a markdown cell and it will render (as shown below). Note that the generated strings can be very lengthy.

```python
## GIFs
import base64

with open("schelling_ani.gif", "rb") as f:    
    b64 = base64.b64encode(f.read()).decode()
print(f'<img src="data:image/png;base64,{b64}">')
```

```python
## PNGs
import base64

with open("schelling_ani.png", "rb") as f:    
    b64 = base64.b64encode(f.read()).decode()
print(f'<img src="data:image/png;base64,{b64}">')
```
